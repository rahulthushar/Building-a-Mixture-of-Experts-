```python
from transformers import pipeline, set_seed
```

### Experiment 1 :: text generation

```python
models = {
    "BERT": "bert-base-uncased",
    "RoBERTa": "roberta-base",
    "BART": "facebook/bart-base"
}

prompt = "The future of Artificial Intelligence is"

for name, model_id in models.items():
    print(f"\n--- {name} ---")
    try:
        generator = pipeline("text-generation", model=model_id)
        output = generator(prompt, max_length=50)
        print(output[0]["generated_text"])
    except Exception as e:
        print("Error:", e)
```

**OBSERVATION :**

-> Experiment 1 demonstrates that encoder-only models such as BERT and RoBERTa are not suitable for text generation. 

-> When forced to generate text, BERT produces repetitive and incoherent output, while RoBERTa fails to extend the prompt meaningfully.        BART, an encoder–decoder model, is able to generate text, though the output lacks coherence due to the absence of task-specific fine-       tuning. 

-> This highlights the importance of architectural design in generative tasks.

### EXPERIMENT 2 :: FILL - MASK

```python
from transformers import pipeline

masked_sentences = {
    "BERT": ("bert-base-uncased",
             "The goal of Generative AI is to [MASK] new content."),
    "RoBERTa": ("roberta-base",
                "The goal of Generative AI is to <mask> new content."),
    "BART": ("facebook/bart-base",
             "The goal of Generative AI is to <mask> new content.")
}

for name, (model_id, sentence) in masked_sentences.items():
    print(f"\n--- {name} ---")
    try:
        fill = pipeline("fill-mask", model=model_id)
        preds = fill(sentence)
        for p in preds[:3]:
            print(p["token_str"], "-", round(p["score"], 4))
    except Exception as e:
        print("Error:", e)
```

-**Observation on fill-mask:**

-  **BERT and RoBERTa** performed strongly on the masked language modeling task, accurately predicting contextually appropriate verbs such as *create* and *generate*.
- **RoBERTa** demonstrated improved contextual understanding by assigning high probability to multiple valid predictions rather than a single dominant choice.
- **BART** produced reasonable but lower-confidence predictions, indicating that masked language modeling is not its primary training objective and highlighting the impact of model architecture and pretraining strategy.


### Experiment 3 : Question Answering

```python
context = 'Generative artificial intelligence, also known as generative AI or GenAI, is a subfield of artificial intelligence that uses generative models to generate text, images, videos, audio, software code or other forms of data.[1] These models learn the underlying patterns and structures of their training data and use them to generate new data[2] in response to input, which often takes the form of natural language prompts.[3][4]'

question = "What are the risks of generative AI?"

for name, model_id in models.items():
    print(f"\n--- {name} ---")
    try:
        qa = pipeline("question-answering", model=model_id)
        result = qa(question=question, context=context)
        print("Answer:", result["answer"])
        print("Score:", round(result["score"], 4))
    except Exception as e:
        print("Failed:", e)
```

-**Observation Q&A:**
-  **BERT** produced an incorrect or vague answer with a very low confidence score, as the base model is not fine-tuned for extractive question answering tasks.
- **RoBERTa** also failed to provide a meaningful answer, showing extremely low confidence due to missing QA-specific output layers and lack of task fine-tuning.
- **BART** initialized missing QA head parameters and produced unreliable results, demonstrating that without downstream fine-tuning, even encoder–decoder models perform poorly on question answering tasks.
