2. ***Strings vs. Messages (Critical Thinking)***

Most people start by talking to the AI like a human: llm.invoke("Translate this to French: Hello")

But LLMs understand Roles:

-System: God-mode instructions. (e.g., "You are a calculator.")  
-Human: The user.   
-AI: The assistant.   

```python
from langchain_core.messages import SystemMessage, HumanMessage

messages = [
    SystemMessage(content="You are a rude teenager, you use slang and don't care"),
    HumanMessage(content="Where is golden temple")
]

response = llm.invoke(messages)
print(response.content)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

if "GOOGLE_API_KEY" not in os.environ:
  os.environ["GOOGLE_API_KEY"] = getpass.getpass("Enter your GOOGLE API key: ")

llm = ChatGoogleGenerativeAI(model = "gemini-2.5-flash")
template = ChatPromptTemplate.from_template(
    "You are a history major. Explain the following topic brefly  in 15 lines:\n\n{topic}"
)

# Output Parser
parser = StrOutputParser()

# LCEL Chain
chain = template | llm | parser

# Invoke
response = chain.invoke({"topic": "Causes of World War 1"})
print(response)
```

### Dynamic Few-Shotting ###

If you have 1000 examples, you can't fit them all in the context window. We use a Selector to pick the best ones.

The Selector Flow (Flowchart)

graph LR
    -Input[User Input] -->|Semantic Search| Database[Example Database]    
    -Database -->|Top 3 Matches| Selector    
    -Selector -->|Inject| Prompt   

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, FewShotChatMessagePromptTemplate
import os, getpass

examples = [
    {"input" : "whats your name? " , "output" : "ning yak bek adu " },
    {"input" : "how is your life ", "output" : "namdhu ond bala "}
]
example_prompt = ChatPromptTemplate.from_messages([
    ("human","{input}"),
    ("ai" , "{output}")
])

few_shot_prompt = FewShotChatMessagePromptTemplate(
    example_prompt = example_prompt,
    examples = examples
)

final_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a dumb 13 your old girl who just says as it is ."),
    few_shot_prompt,
    ("human", "{text}")
])

# Chain
chain = final_prompt | llm

# Run
print(chain.invoke(input("Enter the question: " )).content)
```
