### Project Idea News article summariser ###   
**Working**    
- User uploads the text directly to terminal or can also upload it through a text file     
- Then the text is processed through 'distilbart' for summarization
- Model generates a 200-300 word summary of the given article



```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import sys

# ----------------------------
# Load model & tokenizer
# ----------------------------
MODEL_NAME = "sshleifer/distilbart-cnn-12-6"
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

# ----------------------------
# Summarization function
# ----------------------------
def summarize_article(article_text, max_len=300, min_len=200):
    inputs = tokenizer(
        article_text,
        return_tensors="pt",
        truncation=True,
        max_length=1024
    )

    summary_ids = model.generate(
        inputs["input_ids"],
        max_length=max_len,
        min_length=min_len,
        do_sample=False
    )

    return tokenizer.decode(summary_ids[0], skip_special_tokens=True)

# ----------------------------
# Input handler function
# ----------------------------
def get_article_from_user():
    print("\nChoose input method:")
    print("1. Paste article in terminal")
    print("2. Load article from text file")

    choice = input("\nEnter choice (1/2): ").strip()

    if choice == "1":
        print("\nPaste your article below.")
        print("Press ENTER on an empty line to finish:\n")

        lines = []
        while True:
            line = input()
            if line.strip() == "":
                break
            lines.append(line)

        return "\n".join(lines)

    elif choice == "2":
        file_path = input("\nEnter path to text file: ").strip()
        try:
            with open(file_path, "r", encoding="utf-8") as f:
                return f.read()
        except Exception as e:
            print("❌ Error reading file:", e)
            exit(1)

    else:
        print("❌ Invalid choice")
        exit(1)

# ----------------------------
# Main execution
# ----------------------------
if __name__ == "__main__":
    article_text = get_article_from_user()
    summary = summarize_article(article_text)

    print("\n" + "="*40)
    print("SUMMARY")
    print("="*40)
    print(summary)
```

## Project: News Article TL;DR Generator

**Category:** Text Summarization (Productivity Use Case)

**Objective:**  
Automatically convert long news articles into a short eight to ten-sentence summary.

**Methodology:**  
A pretrained abstractive summarization model is applied to compress lengthy news content into a concise format.

**Model Selected:**  
sshleifer/distilbart-cnn-12-6

**Justification for Model Selection:**  
DistilBART is optimized for speed and efficiency, making it well-suited for rapid summarization tasks.

The system generated compact summaries that retained the essential information from the original articles.
